```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
data=pd.read_csv(r"C:\Users\user\Downloads\archive (1)\heart_statlog_cleveland_hungary_final.csv")
print(data.head())
```

```python
print(data.shape)
print(data.columns)
print(data.isnull().sum())
```

```python
X=data.drop("target", axis=1)
y=data["target"]

X=pd.get_dummies(X, drop_first=True, dtype=int)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)

scaler=StandardScaler()

X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
accuracies={}

for k in [3, 5, 7, 9]:

    model=KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    prediction=model.predict(X_test)
    accuracy=accuracy_score(y_test, prediction)
    accuracies[k]=accuracy

    print("K=", k)
    print("Accuracy =", accuracy*100, "%")
```

```python
best_k=max(accuracies, key=accuracies.get)

print("Best K Value:", best_k)
print("Best Accuracy:", accuracies[best_k]*100, "%")
```


# Conceptual Questions and Answers

## Q1. What is KNN?

K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for classification and regression. It predicts the result of a new data point by looking at the nearest data points in the dataset.

## Q2. How does KNN classify a new data point?

KNN calculates the distance between a new data point and existing data points. It finds the K nearest neighbors and assigns the class that appears most frequently among them.

## Q3. What is the role of K in KNN?

K represents the number of nearest neighbors considered when making a prediction. For example, if K = 3, the algorithm checks the three closest data points to decide the result.

## Q4. Why is feature scaling necessary for KNN?

Feature scaling is necessary because KNN uses distance calculations. Features with larger values can affect the distance more than smaller values. Scaling brings all features to a similar range and helps the algorithm make better predictions.

## Q5. What is Euclidean Distance?

Euclidean distance is the straight-line distance between two data points. KNN uses it to find the nearest neighbors.

**Formula:**

d = √[(x1-y1)² + (x2-y2)² + ... + (xn-yn)²]

## Q6. How can changing K affect model predictions?

Changing K can affect the prediction of the model.

- Small K: The model is more sensitive to noise and individual data points.
- Large K: The model considers more neighbors and produces smoother predictions.

The best K value depends on the dataset and model accuracy.
